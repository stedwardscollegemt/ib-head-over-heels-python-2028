### Linked Lists (Abstract Data Structure)

##### Singly Linked List

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None
```

```python
head = Node(6)
head.next = Node(8)
head.next.next = Node(48)
head.next.next.next = Node(10)
```

```python
print(head)
print(head.next)
print(head.next.next)
print(head.next.next.next)
```
