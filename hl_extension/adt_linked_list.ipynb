### Linked Lists (Abstract Data Structure)

A linked list is a basic but important *linear* data structure where elements, each element is often called a `Node`, are **not** stored at contiguous locations in computer memory.

Instead, each `Node` contains two basic parts:
1. **Data**: The actual value or payload you want to store (e.g., a number, text, or an object).
2. **Link**: A reference to the memory location of the `next` element in the sequence.

##### Singly Linked List (Practical Example)

**How does it work?**

- **The Head**: The first node in a linked list is known as the head. To read or traverse the list, your program starts at the head and follows the pointers step-by-step.
- **The Tail**: The very last node in the list points to `None`, signaling that the list has ended.

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None
```

```python
head = Node(6)
head.next = Node(8)
head.next.next = Node(48)
head.next.next.next = Node(10)
```

```python
print(head)
print(head.next)
print(head.next.next)
print(head.next.next.next)
```

```python
# traverse through the linked list and print out the data stored

current = head
while current:
    print(current.data, end= "--->")
    current = current.next
print("None")
```

```python
# perform the linear search to find the target 48

target = 48
current = head
search = True
while current and search:
    if current.data == target:
        print("Search successful!", current)
        search = False
    else:
        current = current.next
```

##### Doubly Linked List (Practical Example)

**How does it work?**

- **Two Directions**: Each `Node` has two pointers, one pointing to the `next` `Node` and one pointing to the `previous` `Node` - thus allowing you to traverse both forward and backward.

```python
# rewrite the Node class

# @Author: Valentino Basi
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None # Points to the next node
        self.prev = None # Points to the previous node
```

```python
# create a doubly linked list to store elements 6, 8, 48, and 10

# @Author: Valentino Basi
# Creating the nodes
head = Node(6)
second = Node(8)
third = Node(48)
tail = Node (10)

# Linking the nodes forward (.next) and backward (.prev)
head.next = second
second.prev = head

second.next = third
third.prev = second

third.next = tail
tail.prev = third
```

```python
# traverse through the linked list in both directions to print out the data stored

# @Author: Valentino Basi
# Forward traversal (from head to tail)
print("Forward traversal:")
current = head
while current:
    print(current.data, end=" <----> ")
    current = current.next
print("None")

# Backward traversal (from tail to head)
print("\n Backward traversal:")
current = tail
while current:
    print(current.data, end= " <----> ")
    current = current.prev
print("None")
```

```python
# perform the linear search to find the target 8

# @Author: Valentino Basi
target = 8
current = head
search = True
while current and search:
    if current.data == target:
        print("\nSearch successful!", current, "Data:", current.data)
        search = False
    else:
        current = current.next
```
