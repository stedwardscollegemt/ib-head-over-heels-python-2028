#### Python Array (Abstract Data Structure)

Array is a **collection** of elements stored at **contiguous memory locations**, used to hold multiple values of the same data type. Python does not have a built-in array type like some other languages, but it does give us an Abstract Data Type to work with.

---

**IB Computer Science Learning Outcome:** 

B2.2.2 – Intro to one-dimensional (1D) arrays in Python.

```python
import array
```

```python
# Create a simple array of integers
my_array = array.array('i', [1, 2, 3, 4, 5])
```

```python
# Access the first element (similar to the list)
print(my_array[0])
```

```python
# Print all elements (same as we do in a list)
for num in my_array:
    print(num)
```

```python
# Array is an Abstract Data Type aka Class
print(type(my_array))
```
