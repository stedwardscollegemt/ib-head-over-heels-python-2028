### Homework: Crazy Eights Simulation

#### Problem scenario

Card games often require players to follow rules when deciding which cards they can play. In Crazy Eights, players try to discard all their cards by matching the suit or rank of a card already in play.

You will create a simplified simulation of this game using Python.

#### Criteria

- Randomly select a card from the deck to represent the card currently in play.
- Randomly select a card to represent the player's card.
- Display both cards to the player.
- Check whether the player's card matches the card in play by suit or rank.
- Output whether the player can discard their card.

For this simplified version, you may ignore the special rules for eights and assume that the two cards selected are different cards from the deck.

```python
#TODO: import the random module

import random
```

```python
#TODO: model the deck using lists to store suits, faces and numbered 

# I am creating lists to define the variables indicating the different characteristics of the cards (suits, faces and numbered).
suits = ["hearts", "spades", "clubs", "diamonds"]
faces = ["ace", "jack", "queen", "king"]
numbered = [2, 3, 4, 5, 6, 7, 8, 9, 10]

# I am hence, using a nested loop, getting the program to create a 52-card deck based on the variables that I have defined above (suits, faces and numbered).
# I am creating a new variable called rank  that merges all the items in the "faces" and "numbered" separate lists into one large single list.
# The new "rank" single list was created since the program must choose the rank (number or face) of a card out of all the possible choices of number and faces, given that a card can only either be numbered or be a face and it cannot be both.


ranks = faces + numbered
the_deck = [{"rank": rank, "suit": suit} for rank in ranks for suit in suits]



# In this last line of code above, I have instructed the program to go through each value defined as a rank (Ace, 2, 3, 4, 5, 6, 7, 8, 9, 10, Jack, Queen, King) and pair each of them to each value defined as a suit (hearts, spades, clubs, diamonds).
# I have used curled brackets in my for loop to define rank and suit as keys (labels) and attribute values to them. Although I have attribuited the values rank and suit respectively to each label, I could have used any value (such as x and y).
# Creating labels for rank and suit do not affect the for loop, but they will be necessary later when comparing the ranks and suits of the two cards.
# This is so as it allows the program to identify "rank" and "suit" as distinct values of a card.



```

```python
# THIS CELL IS FOR TESTING PURPOSES ONLY - PLEASE IGNORE (IT IS NOT AN INTENDED PART OF THE ASSIGNMENT OR USER EXPERIENCE BUT HAS BEEN INCLUDED FOR COMPLETENESS).
for card in the_deck:
    print(card)

print(len(the_deck))

# Obtaining an output of 52 for the length (len) of the deck and seeing each possible card printed confirms that the deck has been created successfully behind the scenes.

```

```python
#TODO: select a random card, select a card for the player, display them both

# I am instructing the program to pick two cards from the deck without replacement (I am using random.sample instead of random.choice) so that the same card cannot be picked twice, as this would be impossible in real life.
# The two cards will be picked in a specific order: the first card picked will be a random one, the second one will be the player's card.
# I am using k=2 to specify that I want two items chosen at random simultaneously (without having to run the program twice).
card_in_play, player_card = random.sample(the_deck, k=2)


# I am instructing the program to output (display to the user) which two cards were picked, which card is the player's card and which one is the card in play.
# I am using an f string (f") so that the output will read as neat continuous text (the program will replace the variable name with actual text when printing the output).

print(f"Card in play: {card_in_play['rank']} of {card_in_play['suit']}")
print(f"Player card: {player_card['rank']} of {player_card['suit']}")
```

```python
#TODO: check whether the player's card matches the suit or rank.
#      if so, display that the card can be discarded

# I am instructing the program to compare the ranks and the suits of the two cards drawn simultaneously, to see if either the ranks or the suits of both cards match.
# I am using or instead of elif since I want both conditions to be evaluated at the same time, rather than evaluating one first and only evaluating the second one if the first one failed.
# I am then instructing the program to provide two different outputs, depending on whether a card may be discarded or not based on the results of the comparison.
# I am using an f string (f") so that the output will read as neat continuous text (the program will replace the variable name with actual text when printing the output).
 
if player_card['rank'] == card_in_play['rank'] or player_card['suit'] == card_in_play['suit']:
    print(f"The player may discard their card, {player_card['rank']} of {player_card['suit']}.")

else: 
    print("No card may be discarded by the player.")
```
