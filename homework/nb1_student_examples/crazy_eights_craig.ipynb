### Homework: Crazy Eights Simulation

#### Problem scenario

Card games often require players to follow rules when deciding which cards they can play. In Crazy Eights, players try to discard all their cards by matching the suit or rank of a card already in play.

You will create a simplified simulation of this game using Python.

#### Criteria

- Randomly select a card from the deck to represent the card currently in play.
- Randomly select a card to represent the player's card.
- Display both cards to the player.
- Check whether the player's card matches the card in play by suit or rank.
- Output whether the player can discard their card.

For this simplified version, you may ignore the special rules for eights and assume that the two cards selected are different cards from the deck.

```python
# import the random module
import random
```

```python
# model the deck using lists to store suits, faces and numbered 

suits = ["Hearts", "Diamonds", "Clubs", "Spades"]
faces = ["Jack", "Queen", "King", "Ace"]
numbers = [2, 3, 4, 5, 6, 7, 8, 9, 10]

ranks = faces + numbers
```

```python
# select a random card from the deck, select a random card for the player, display them both

play_suit = random.choice(suits)
play_rank = random.choice(ranks)

player_suit = random.choice(suits)
player_rank = random.choice(ranks)

while play_suit == player_suit and play_rank == player_rank:
    player_suit = random.choice(suits)
    player_rank = random.choice(ranks)

print("Card in play:", play_rank, "of" , play_suit) 
print("Player's card:", player_rank, "of", player_suit)   
```

```python
# check whether the player's card matches the suit or rank.
# if so, display that the card can be discarded

if play_suit == player_suit or play_rank == player_rank:
    print("You can discard your card!")
else:
    print("You cannot discard your card!")
```
