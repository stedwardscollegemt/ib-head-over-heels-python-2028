```python
import random

def draw():
    the_suit = random.choice(suits)
    the_card = random.choice(faces + numbered)
    return the_card, "of", the_suit

suits = ["Clubs", "Spades", "Hearts", "Diamonds"]
faces = ["Jack", "Queen", "King", "Ace"]
numbered = [2, 3, 4, 5, 6, 7, 8, 9, 10]

card_in_play = draw()
player_card = draw()

print("card_in_play",card_in_play)
print("player_card",player_card)

if player_card[0] == card_in_play[0] or player_card[2] == card_in_play[2]:
    print("The card can be discarded.")
else:
    print("The card cannot be discarded.")
```
