### Homework: Electricity Consumption Calculator

#### Problem scenario

People often need help with estimating the electricity being consumed by their home appliances. 

You can help them by writing a small program that calculates the electricity used by an appliance.

#### Criteria

Your program should enable to the user to enter:
- the power of the appliance in watts
- how many hours it is used each day
- the number of days it is used

The program should calculate the total energy consumption in kilowatt-hours (kWh) using the following formula.

```
energy = (power × hours × days) / 1000
```

Finally, based on the energy calculated, the program output whether their appliance has low, medium, or high consumption.

````python
# Write a function that calculates the total energy consumption using the formula

def total_energy(power: float, hours: float, days: float) -> float:
    return (power * hours * days)/1000

# I have defined the components in the formula to calculate the total energy (power, days and hours) and I have written a formula establishing how they are used to find it.
````

````python
# Ask the user to input the appliance's power, hours of use per day and number of days 

power_in = float(input("Enter the power (in watts) used by the appliance:"))
hours_in = float(input("Enter the number of hours the appliance is used for each day:"))
days_in = float(input("Enter the number of days the appliance is used for:"))

# I am programming the computer to request the user to input the values for power, hours and days needed for the calculation.
````

````python
# Call the function to get the result in kWh

average_energy_consumption = total_energy(power_in, hours_in, days_in)

# I am naming the result of the total energy calculation.
````

````python
# Write selection statements so that based on the result, the program outputs "Low", "Medium" or "High"
#      - Low if consumption is less than 10 kWh
#      - Medium if consumption is between 10 and 30 kWh
#      - High if consumption is greater than 30 kWh

if power_in < 0 or days_in < 0 or hours_in < 0:
    print("Fail")
elif average_energy_consumption < 10:
    print("Low")
elif average_energy_consumption >= 10 and average_energy_consumption <= 30: 
    print("Medium")
elif average_energy_consumption > 30:
    print("High")

# I am establishing conditions (intervals) that potential calculation valued obtained must lie in to be classified in a certain manner. 
# I am also excluding any negative input value for days, hours and power since they cannot logically exist in this context.
````
