### Homework: Electricity Consumption Calculator

#### Problem scenario

People often need help with estimating the electricity being consumed by their home appliances. 

You can help them by writing a small program that calculates the electricity used by an appliance.

#### Criteria

Your program should enable to the user to enter:
- the power of the appliance in watts
- how many hours it is used each day
- the number of days it is used

The program should calculate the total energy consumption in kilowatt-hours (kWh) using the following formula.

```
energy = (power × hours × days) / 1000
```

Finally, based on the energy calculated, the program output whether their appliance has low, medium, or high consumption.

````python
#TODO: Write a function that calculates the total energy consumption using the formula

def calculate_energy(power, hours, days):
    energy = (power * hours * days) / 1000
    return energy
````

````python
#TODO: Ask the user to input the appliance's power, hours of use per day and number of days 

power = float(input("Enter the power of the appliance in watts: "))
hours = float(input("Enter the ammount of hours the appliance is used per day: "))
days = int(input("Enter the number of days used: "))
````

````python
#TODO: Call the function to get the result in kWh

total_energy = calculate_energy(power, hours, days)
print("Total energy consuption:" , total_energy , "kWh")
````

````python
#TODO: Write selection statements so that based on the result, the program outputs "Low", "Medium" or "High"
#      - Low if consumption is less than 10 kWh
#      - Medium if consumption is between 10 and 30 kWh
#      - High if consumption is greater than 30 kWh

if total_energy < 10:
    print("Low consumption")

elif total_energy <= 30:
    print("Medium consumption")

else:
    print("High consumption")    
````
